## Task 4: In-place operators

## Task 14: Simple fully connected classifier

The simplest classifier for MNIST often uses flattening, one hidden layer, ReLU, and a final output layer.


```python
class FCClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 128),
            nn.ReLU(),
            nn.Linear(128, 10),
        )

    def forward(self, x):
        return self.net(x)
```

## Task 15: Loss function and optimizer

Classification models usually use cross-entropy loss and an optimizer such as Adam.


```python
model = FCClassifier()
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

## Task 16: Loading MNIST with torchvision

The torchvision dataset utilities can download MNIST and apply a transform automatically.


```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform,
)

print(len(train_dataset))
```

```python


train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)
```

## Task 20: Training and evaluation workflow

Training updates the weights, while evaluation measures performance without tracking gradients.


```python
for epoch in range(5):
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

model.eval()
with torch.no_grad():
    outputs = model(images)
    _, predicted = torch.max(outputs, 1)
```
